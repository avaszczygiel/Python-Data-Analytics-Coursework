## Loading the Pre-Trained Model

We will use **GloVe** (Global Vectors for Word Representation), a model trained on Wikipedia and news articles. It knows about 400,000 words.

We use the 50-dimensional version for speed. Production models typically use 100–300 dimensions.

**This download takes 1–2 minutes the first time.** It will be cached after that.

```python
# Download the pre-trained GloVe model (50 dimensions, ~66 MB)
# This takes 1-2 minutes the first time
model = api.load('glove-wiki-gigaword-50')
print(f"Model loaded! Vocabulary size: {len(model)} words")
print(f"Vector size: {model.vector_size} dimensions")
```

---
## What Does a Word Vector Look Like?

Each word is represented as a list of 50 numbers. These numbers don't have obvious meaning on their own, but the **pattern** across all 50 numbers encodes what the word means.

```python
# Look at the vector for "cat"
cat_vector = model['cat']
print(f"The word 'cat' is represented by {len(cat_vector)} numbers:")
print(cat_vector[:10])  # Show first 10 numbers
print(f"... (and {len(cat_vector) - 10} more)")
print()

# Compare: "cat" and "kitten" have similar vectors
kitten_vector = model['kitten']
print("First 5 numbers of 'cat':   ", np.round(cat_vector[:5], 2))
print("First 5 numbers of 'kitten':", np.round(kitten_vector[:5], 2))
print()

# Compare: "cat" and "democracy" have very different vectors
democracy_vector = model['democracy']
print("First 5 numbers of 'democracy':", np.round(democracy_vector[:5], 2))
```

## Demo: Finding Similar Words

The model can tell us which words are most similar to a given word. The similarity score ranges from 0 (unrelated) to 1 (identical).

```python
# What words are most similar to "cat"?
print("Words most similar to 'cat':")
for word, score in model.most_similar('cat', topn=5):
    print(f"  {word:15s} {score:.3f}")

print()

# Compare specific pairs
print(f"Similarity('cat', 'kitten'):    {model.similarity('cat', 'kitten'):.3f}")
print(f"Similarity('cat', 'dog'):       {model.similarity('cat', 'dog'):.3f}")
print(f"Similarity('cat', 'democracy'): {model.similarity('cat', 'democracy'):.3f}")
```

## Demo: Word Arithmetic

Because vectors encode meaning, you can do **math with words**:
- king - man + woman = queen

This works because the vectors capture **relationships** between concepts.

```python
# king - man + woman = ?
print("king - man + woman = ?")
result = model.most_similar(positive=['king', 'woman'], negative=['man'], topn=3)
for word, score in result:
    print(f"  {word:15s} {score:.3f}")

print()

# Paris - France + Italy = ?
print("Paris - France + Italy = ?")
result = model.most_similar(positive=['paris', 'italy'], negative=['france'], topn=3)
for word, score in result:
    print(f"  {word:15s} {score:.3f}")
```

## Activity 2: Document Vectors & Visualization (~20 min)

*Your CEO hands you 6 customer reviews and says: "Group these into similar topics — but don't just match keywords. Two reviews about 'terrible service' and 'awful experience' should end up together even though they use different words."*

### Demo: Creating Document Vectors

To represent an entire document as a single vector, we **average** the vectors of all its words. This gives us one 50-number summary for the whole text.

```python
def doc_vector(text, model):
    """Average the word vectors for all known words in a text."""
    words = text.lower().split()
    vectors = [model[w] for w in words if w in model]
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    return np.mean(vectors, axis=0)

# Example
vec = doc_vector("The cat sat on the mat", model)
print(f"Document vector shape: {vec.shape}")
print(f"First 5 values: {np.round(vec[:5], 3)}")
```

### Demo: Cosine Similarity Between Documents

```python
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "The cat sat on the mat",
    "A kitten rested on the rug",
    "Stock prices rose on Wall Street today"
]

vectors = np.array([doc_vector(d, model) for d in docs])
sim_matrix = cosine_similarity(vectors)

print("Cosine similarity matrix:")
print()
for i, doc in enumerate(docs):
    print(f"  Doc {i+1}: {doc[:40]}...")
print()
print(np.round(sim_matrix, 3))
print()
print(f"Doc 1 vs Doc 2 (cat/kitten): {sim_matrix[0,1]:.3f} -- HIGH (same topic!)")
print(f"Doc 1 vs Doc 3 (cat/stocks): {sim_matrix[0,2]:.3f} -- LOW (different topics)")
```

### 📝 Activity 2a: Document Similarity

Create document vectors for the 6 customer reviews below and compute a cosine similarity matrix. Which reviews are most similar to each other?

Steps:
1. Use the `doc_vector()` function from the demo
2. Create a vector for each review
3. Compute `cosine_similarity()` on the matrix of vectors
4. Display the similarity matrix as a heatmap

```python
import seaborn as sns

reviews = [
    "The food was absolutely delicious and the service was excellent",
    "Terrible experience the waiter was rude and the meal was cold",
    "Great software product with an intuitive user interface",
    "The restaurant had amazing pasta and friendly staff",
    "Buggy application that crashes frequently very frustrating",
    "Outstanding dining with wonderful flavors and attentive waiters"
]

labels = ['R1: good food', 'R2: bad food', 'R3: good tech',
          'R4: good food', 'R5: bad tech', 'R6: good food']

# Steps:
# 1) build review_vectors with doc_vector(...) for each review
# 2) compute the cosine similarity matrix
# 3) display it as a sns.heatmap with the labels above

# YOUR CODE HERE
review_vectors = [doc_vector(r, model) for r in reviews]
sim_matrix = cosine_similarity(review_vectors)

plt.figure(figsize=(10, 8))
sns.heatmap(
    sim_matrix,
    annot=True,
    xticklabels=labels,
    yticklabels=labels,
    cmap='YlGnBu'
)

plt.title('Cosine Similarity Heatmap: Food vs. Tech Reviews')
plt.show()
# Expected: R1, R4, R6 (good food) cluster; R3 and R5 (tech) somewhat similar.
```

### 📝 Activity 2b: Visualize Documents with PCA

Use PCA to reduce the 50-dimensional document vectors to 2D and create a scatter plot. Color the points by category.

Steps:
1. Use `PCA(n_components=2)` from sklearn
2. Transform the review vectors to 2D
3. Create a scatter plot with labels

```python
from sklearn.decomposition import PCA

categories = ['good food', 'bad food', 'good tech',
              'good food', 'bad tech', 'good food']
color_map = {'good food': 'green', 'bad food': 'red',
             'good tech': 'blue', 'bad tech': 'orange'}
colors = [color_map[c] for c in categories]

# Steps:
# 1) PCA(n_components=2) on review_vectors (from 2a)
# 2) scatter plot, color by category, annotate each point with its label

# YOUR CODE HERE
pca = PCA(n_components=2)
reduced_vectors = pca.fit_transform(review_vectors)
plt.figure(figsize=(10, 7))
for i, (x, y) in enumerate(reduced_vectors):
    plt.scatter(x, y, color=colors[i], edgecolors='k', s=100)
    # Annotate with the specific label (R1, R2, etc.)
    plt.text(x + 0.01, y + 0.01, labels[i], fontsize=12)
from matplotlib.lines import Line2D
legend_elements = [Line2D([0],[0], marker='o', color='w', label=k,
                          markerfacecolor=v, markersize=10) for k, v in color_map.items()]
plt.legend(handles=legend_elements, loc='best')

plt.title('PCA Projection of Review Embeddings')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
# Expected: food reviews cluster together; tech reviews cluster together.
```

## Activity 3: Word Embeddings for Business Terms (~15 min)

In this activity you will explore how well the model understands business and industry-specific vocabulary.

### 📝 Activity 3a: Industry Word Clusters

Pick 3 industries (e.g., finance, healthcare, technology). For each, choose 4 words related to that industry. Then:
1. Get the vectors for all 12 words
2. Use PCA to plot them in 2D
3. Do words from the same industry cluster together?

*Hint: follow the same pattern as the word visualization demo in the slides.*

```python
# Activity 3a: Industry word clusters
# Pick 3 industries, 4 words each. Get vectors, reduce to 2D with PCA, scatter with colors and labels.
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
# YOUR CODE HERE
industry_data = {
    'Finance': ['bank', 'investment', 'currency', 'stocks'],
    'Healthcare': ['hospital', 'doctor', 'medicine', 'patient'],
    'Technology': ['software', 'computer', 'digital', 'algorithm']
}
words = []
categories = []
colors = []
color_palette = {'Finance': 'blue', 'Healthcare': 'red', 'Technology': 'green'}
word_vectors = []
for industry, industry_words in industry_data.items():
    for word in industry_words:
        if word in model:
            word_vectors.append(model[word])
            words.append(word)
            categories.append(industry)
            colors.append(color_palette[industry])
        else:
            print(f"Word '{word}' not in model vocabulary.")
pca = PCA(n_components=2)
pca_result = pca.fit_transform(word_vectors)
plt.figure(figsize=(10, 8))

for i, word in enumerate(words):
    x, y = pca_result[i, 0], pca_result[i, 1]
    plt.scatter(x, y, color=colors[i], s=100, edgecolors='k')
    plt.text(x + 0.01, y + 0.01, word, fontsize=12)
from matplotlib.lines import Line2D
legend_elements = [Line2D([0],[0], marker='o', color='w', label=k,
                          markerfacecolor=v, markersize=10) for k, v in color_palette.items()]
plt.legend(handles=legend_elements, title="Industries")

plt.title('Word Cluster Visualization (PCA)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
# Expected: words from the same industry appear near each other.
```

### 📝 Activity 4b: Embeddings vs. TF-IDF

Let's compare how TF-IDF and word embeddings handle synonyms. Create TF-IDF vectors for the 6 customer reviews from Activity 2a, compute cosine similarity, and compare with the embedding-based similarity.

Steps:
1. Use `TfidfVectorizer` from sklearn to vectorize the reviews
2. Compute cosine similarity
3. Display both heatmaps side by side

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
# Same reviews and labels from Activity 2a
reviews = [
    "The food was absolutely delicious and the service was excellent",
    "Terrible experience the waiter was rude and the meal was cold",
    "Great software product with an intuitive user interface",
    "The restaurant had amazing pasta and friendly staff",
    "Buggy application that crashes frequently very frustrating",
    "Outstanding dining with wonderful flavors and attentive waiters"
]

labels = ['R1: good food', 'R2: bad food', 'R3: good tech',
          'R4: good food', 'R5: bad tech', 'R6: good food']

# Steps:
# 1) Build TF-IDF vectors with TfidfVectorizer

# 2) Build embedding vectors with doc_vector
# 3) Compute cosine similarity for both
# 4) Plot two heatmaps side by side and compare

# YOUR CODE HERE
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(reviews)
tfidf_sim = cosine_similarity(tfidf_matrix)

review_vectors = [doc_vector(r, model) for r in reviews]
embedding_sim = cosine_similarity(review_vectors)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

sns.heatmap(tfidf_sim, annot=True, xticklabels=labels, yticklabels=labels,
            cmap='YlGnBu', ax=ax1)
ax1.set_title('TF-IDF Cosine Similarity')

sns.heatmap(embedding_sim, annot=True, xticklabels=labels, yticklabels=labels,
            cmap='YlGnBu', ax=ax2)
ax2.set_title('Word2Vec Embedding Cosine Similarity')

plt.tight_layout()
plt.show()
# Expected: embeddings show higher similarity between R1/R4/R6 (good food)
# even when the wording differs; TF-IDF can miss this because it relies on
# exact-word overlap.
```
