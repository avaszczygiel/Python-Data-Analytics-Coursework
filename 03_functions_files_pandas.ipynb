# Part 1: Creating Functions (15 min) 🔧

Functions are reusable blocks of code that perform specific tasks.

### The DRY Principle: Don't Repeat Yourself!

```python
def function_name(parameter1, parameter2):
    """Docstring: describes what the function does."""
    # Code that does something
    return result
```

| Component | Purpose |
|-----------|--------|
| `def` | Keyword that defines a new function |
| Parameters | Inputs to the function |
| Docstring | Documentation in triple quotes |
| `return` | Output value (optional) |

````python
# EXAMPLE: Basic function anatomy

def greet_customer(name):
    """Generate a greeting message for a customer."""
    message = f"Hello, {name}! Welcome to our store."
    return message

# Call the function
result = greet_customer("Alice")
print(result)

# Or print directly
print(greet_customer("Bob"))
````

````python
# EXAMPLE: Why functions matter - The DRY principle

# WITHOUT functions (repetitive and error-prone):
area1 = 5 * 3
area2 = 7 * 4
area3 = 10 * 2
print(f"Areas (repetitive): {area1}, {area2}, {area3}")

# WITH functions (clean and reusable):
def calculate_area(length, width):
    """Calculate the area of a rectangle."""
    return length * width

print(f"Areas (with function): {calculate_area(5,3)}, {calculate_area(7,4)}, {calculate_area(10,2)}")

# Access the docstring:
print(f"\nDocstring: {calculate_area.__doc__}")
````

````python
# EXAMPLE: Functions with multiple parameters and default values

def create_email(name, domain="company.com", greeting="Hello"):
    """Create an email address and greeting."""
    email = f"{name.lower()}@{domain}"
    message = f"{greeting}, {name}! Your email is {email}"
    return message

# Using all defaults
print(create_email("Alice"))

# Override one default
print(create_email("Bob", "gmail.com"))

# Using keyword arguments (can be in any order!)
print(create_email("Carol", greeting="Welcome", domain="outlook.com"))
````

Create a function that returns multiple values (as a tuple).

````python
# ACTIVITY 1.3: Statistics Calculator (returns multiple values)

def get_stats(numbers):
    """
    Calculate basic statistics for a list of numbers.
    Returns: (total, average, minimum, maximum)
    """
    # YOUR CODE: Calculate total, average, min, max
    total = sum(numbers)
    average = total/len(numbers)
    minimum = min(numbers)
    maximum = max(numbers)

    return total, average, minimum, maximum  # Returns a tuple!

# Test your function:
data = [10, 20, 30, 40, 50]

# Unpack the returned tuple
total, avg, min_val, max_val = get_stats(data)

print(f"Data: {data}")
print(f"Total: {total}")      # Expected: 150
print(f"Average: {avg}")      # Expected: 30.0
print(f"Min: {min_val}")      # Expected: 10
print(f"Max: {max_val}")      # Expected: 50
````

# Part 2: Parameters, Return Values & Scope (15 min) 🎯

### Variable Scope: Where Variables Live

| Scope | Where Defined | Accessible From |
|-------|--------------|----------------|
| **Local** | Inside a function | Only inside that function |
| **Global** | Outside all functions | Everywhere in the file |

> ⚠️ **Rule:** Python searches Local → Global → Built-in

````python
# EXAMPLE: Variable scope

x = "I am GLOBAL"  # Global scope

def my_function():
    y = "I am LOCAL"  # Local scope - only exists inside this function
    print(f"Inside function: x = '{x}'")  # Can read global
    print(f"Inside function: y = '{y}'")  # Can read local

my_function()

print(f"\nOutside function: x = '{x}'")  # Still works
# print(y)  # This would cause an ERROR! y doesn't exist here
````

````python
# EXAMPLE: Lambda functions (quick anonymous functions)

# Regular function:
def square(x):
    return x ** 2

# Lambda equivalent (one-liner):
square_lambda = lambda x: x ** 2

print(f"square(5) = {square(5)}")
print(f"square_lambda(5) = {square_lambda(5)}")

# Common use: sorting with a custom key
students = [("Alice", 85), ("Bob", 92), ("Carol", 78)]

# Sort by score (second element of each tuple)
sorted_students = sorted(students, key=lambda s: s[1])
print(f"\nSorted by score: {sorted_students}")

# Sort by score descending
sorted_desc = sorted(students, key=lambda s: s[1], reverse=True)
print(f"Sorted descending: {sorted_desc}")
````

### 📝 ACTIVITY 2.1: Password Validator

Create a function that validates a password based on multiple criteria.

````python
# ACTIVITY 2.1: Password Validator

def validate_password(password):
    """
    Check if a password meets security requirements:
    - At least 8 characters long
    - Contains at least one digit

    Returns: (is_valid, message)
    """
    # YOUR CODE: Check length
    if len(password) < 8:
        return False, "Password must be at least 8 characters"         # Return False and an error message

    # YOUR CODE: Check for at least one digit
    has_digit = False
    for char in password:
        if char.isdigit():
            has_digit = True
            break

    if not has_digit:
        return False, "Password must contain at least one digit" # Return False and an error message

    return True, "Password is valid!"  # Return True and a success message

# Test passwords:
test_passwords = ["abc", "abcdefgh", "secure123", "12345678"]

for pwd in test_passwords:
    is_valid, message = validate_password(pwd)
    status = "✓" if is_valid else "✗"
    print(f"{status} '{pwd}': {message}")

# Expected:
# ✗ 'abc': Password must be at least 8 characters
# ✗ 'abcdefgh': Password must contain at least one digit
# ✓ 'secure123': Password is valid!
# ✓ '12345678': Password is valid!
````

# Part 4: Series & DataFrames (15 min) 📊

### The Two Core Pandas Data Structures

| Structure | Description | Think of it as... |
|-----------|-------------|------------------|
| **Series** | 1D labeled array | A list with superpowers |
| **DataFrame** | 2D labeled table | A dict of Series (columns) |

> 💡 Every column in a DataFrame is a Series!

````python
# EXAMPLE: Series - A list with superpowers

# Plain Python list - limited capabilities
ages_list = [25, 30, 35, 40]
# ages_list.mean()  # This would ERROR!

# Wrap it in a Series - get superpowers!
ages_series = pd.Series([25, 30, 35, 40])
print("=== Series ===")
print(ages_series)

print("\n=== Series Superpowers ===")
print(f"Mean: {ages_series.mean()}")
print(f"Sum: {ages_series.sum()}")
print(f"Max: {ages_series.max()}")
print(f"Multiply by 2: \n{ages_series * 2}")
````

````python
# EXAMPLE: Selecting columns

# Single column (returns a Series)
print("=== Single column (Series) ===")
print(df['name'])

# Multiple columns (returns a DataFrame)
# NOTE: Double brackets [[ ]] !
print("\n=== Multiple columns (DataFrame) ===")
print(df[['name', 'salary']])

# Common mistake:
# df['name', 'salary']  # WRONG!
# df[['name', 'salary']] # CORRECT - list inside brackets!
````

# Part 5: Selecting Data with .loc[] and .iloc[] (15 min) 🎯

### Quick Comparison

| Feature | `.loc[]` | `.iloc[]` |
|---------|----------|----------|
| Uses | Index values + column **names** | Position **numbers** only |
| Slicing | **Inclusive**: `[0:2]` gives 0, 1, 2 | **Exclusive**: `[0:2]` gives 0, 1 |
| Best for | When you know column names | When you know position numbers |

> ⚠️ The slicing behavior is the KEY difference: `.loc[]` includes the endpoint, `.iloc[]` excludes it (like Python lists).

````python
# EXAMPLE: .loc[] - uses labels/names

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Carol', 'David', 'Eve'],
    'age': [25, 30, 35, 40, 45],
    'salary': [50000, 60000, 70000, 80000, 90000]
})

print("=== DataFrame ===")
print(df)

print("\n=== .loc[] Examples ===")

# Single row by index
print(f"df.loc[0] (row 0):\n{df.loc[0]}\n")

# Single cell: row 0, column 'name'
print(f"df.loc[0, 'name']: {df.loc[0, 'name']}")

# Slice of rows (INCLUSIVE!)
print(f"\ndf.loc[0:2] (rows 0, 1, AND 2):")
print(df.loc[0:2])
````

````python
# EXAMPLE: .iloc[] - uses position numbers only

print("=== .iloc[] Examples ===")

# First row
print(f"df.iloc[0] (first row):\n{df.iloc[0]}\n")

# Last row
print(f"df.iloc[-1] (last row):\n{df.iloc[-1]}\n")

# Slice of rows (EXCLUSIVE - like Python!)
print(f"df.iloc[0:2] (rows 0 and 1 only, NOT 2):")
print(df.iloc[0:2])

# Row 0, column 1 (by position)
print(f"\ndf.iloc[0, 1] (row 0, column 1): {df.iloc[0, 1]}")
````

````python
# EXAMPLE: More selection patterns

print("=== Advanced Selection ===")

# All rows, specific columns with .loc[]
print("df.loc[:, ['name', 'salary']] (all rows, 2 columns):")
print(df.loc[:, ['name', 'salary']])

# Multiple specific rows
print("\ndf.loc[[0, 2, 4]] (rows 0, 2, and 4):")
print(df.loc[[0, 2, 4]])
````

# Part 6: Boolean Indexing & Modifying Data (10 min) 🔍

### Boolean Indexing = Filtering Rows

```python
# Step 1: Create a condition (returns True/False for each row)
df['age'] > 30

# Step 2: Use condition to filter
df[df['age'] > 30]
```

### Combining Conditions

| Operator | Python | Pandas |
|----------|--------|--------|
| AND | `and` | `&` |
| OR | `or` | `\|` |
| NOT | `not` | `~` |

> ⚠️ **Important:** Each condition MUST be in parentheses!

````python
# EXAMPLE: Boolean indexing

employees = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Carol', 'David', 'Eve'],
    'department': ['IT', 'HR', 'IT', 'Sales', 'IT'],
    'salary': [75000, 55000, 82000, 68000, 91000],
    'years': [3, 7, 5, 2, 8]
})

print("=== All Employees ===")
print(employees)

# Single condition: salary > 70000
print("\n=== Salary > $70,000 ===")
print(employees[employees['salary'] > 70000])

# Multiple conditions: IT department AND salary > 80000
print("\n=== IT AND Salary > $80,000 ===")
print(employees[(employees['department'] == 'IT') & (employees['salary'] > 80000)])
````

````python
# EXAMPLE: Adding and modifying columns

# Add a new column (calculated)
employees['bonus'] = employees['salary'] * 0.1
print("=== With Bonus Column ===")
print(employees)

# Add a boolean column
employees['senior'] = employees['years'] >= 5
print("\n=== With Senior Column ===")
print(employees)
````

### Scenario: Sales Analysis Report

You're a data analyst at a retail company. Use everything you've learned to analyze the sales data and create a summary report.

````python
# CAPSTONE CHALLENGE: Sales Analysis

# Company sales data
sales_data = pd.DataFrame({
    'rep_name': ['Alice', 'Bob', 'Carol', 'David', 'Eve', 'Frank'],
    'region': ['North', 'South', 'North', 'East', 'South', 'East'],
    'q1_sales': [45000, 38000, 52000, 41000, 48000, 35000],
    'q2_sales': [48000, 42000, 55000, 39000, 51000, 38000],
    'target': [90000, 75000, 100000, 85000, 95000, 70000]
})

print("=" * 60)
print(f"{'SALES ANALYSIS REPORT':^60}")
print("=" * 60)

# ============================================
# Task 1: Add calculated columns
# ============================================

# Add 'total_sales' column (q1 + q2)
sales_data['total_sales'] = sales_data['q1_sales'] + sales_data['q2_sales']   # YOUR CODE

# Add 'met_target' column (True if total_sales >= target)
sales_data['met_target'] = sales_data['total_sales'] >= sales_data['target']   # YOUR CODE

# Add 'pct_of_target' column (total_sales / target * 100)
sales_data['pct_of_target'] = (sales_data['total_sales'] / sales_data['target']) * 100 # YOUR CODE

print("\n=== Sales Data with Calculated Columns ===")
print(sales_data)

# ============================================
# Task 2: Create a function for the analysis
# ============================================

def analyze_sales(df):
    """
    Analyze sales data and return key metrics.
    Returns: (total_revenue, avg_per_rep, reps_met_target)
    """
    # YOUR CODE: Calculate metrics
    total_revenue = df['total_sales'].sum()    # Sum of all total_sales
    avg_per_rep = df['total_sales'].mean()     # Average of total_sales
    reps_met_target = df['met_target'].sum()  # Count of True values in met_target

    return total_revenue, avg_per_rep, reps_met_target

# Call the function
total, avg, met = analyze_sales(sales_data)

print("\n=== Summary Statistics ===")
print(f"Total Revenue: ${total:,}")
print(f"Average per Rep: ${avg:,.2f}")
print(f"Reps Meeting Target: {met} of {len(sales_data)}")

# ============================================
# Task 3: Filter and identify top performers
# ============================================

print("\n=== Top Performers (Met Target) ===")
top_performers = sales_data[sales_data['met_target'] == False] # YOUR CODE: Filter where met_target is True
print(top_performers[['rep_name', 'region', 'total_sales', 'pct_of_target']])

print("\n=== Needs Improvement (Below Target) ===")
needs_improvement = sales_data[sales_data['met_target'] == False]  # YOUR CODE: Filter where met_target is False
print(needs_improvement[['rep_name', 'region', 'total_sales', 'pct_of_target']])

print("\n" + "=" * 60)
print("Report Complete!")
print("=" * 60)
````

### Common Mistakes to Avoid:

1. **Forgetting `return`**: Functions without `return` give `None`
2. **Single vs double brackets**: `df['col']` (Series) vs `df[['col']]` (DataFrame)
3. **Slicing differences**: `.loc[0:2]` = 3 rows, `.iloc[0:2]` = 2 rows
4. **Missing parentheses in filters**: `df[(cond1) & (cond2)]` NOT `df[cond1 & cond2]`
5. **Using `and` instead of `&`**: Pandas uses `&`, `|`, `~` NOT `and`, `or`, `not`
