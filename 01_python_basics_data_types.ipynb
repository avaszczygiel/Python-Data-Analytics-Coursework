# Part 2: Numeric Types & Booleans (15 min) 🔢

### Numeric Types

| Type | Description | Examples |
|------|-------------|----------|
| `int` | Whole numbers | `42`, `-17`, `0` |
| `float` | Decimal numbers | `3.14`, `-0.5`, `2.0` |

### Booleans (`bool`)

Only two values: `True` and `False`. They result from comparisons and are the basis of all logic in Python.

**Falsy values** (evaluate to `False`): `0`, `0.0`, `""`, `[]`, `()`, `{}`, `None`

**Truthy values** (evaluate to `True`): Everything else!

````python
# EXAMPLE: Numeric operations and division behavior

a = 10
b = 3

print("=== Arithmetic ===")
print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
print(f"{a} ** {b} = {a ** b}")   # Exponentiation

print("\n=== Division (watch the types!) ===")
print(f"{a} / {b}  = {a / b}   type: {type(a / b)}")    # Always returns float!
print(f"{a} // {b} = {a // b}    type: {type(a // b)}")  # Floor division returns int
print(f"{a} % {b}  = {a % b}    (remainder)")
````

````python
# EXAMPLE: Booleans and truthy/falsy

x = 10
print("=== Comparisons ===")
print(f"x > 5:  {x > 5}")
print(f"x == 5: {x == 5}")
print(f"x != 5: {x != 5}")

print("\n=== Truthy / Falsy ===")
print(f"bool(1):       {bool(1)}")
print(f"bool(0):       {bool(0)}")
print(f"bool('hello'): {bool('hello')}")
print(f"bool(''):      {bool('')}")
print(f"bool([]):      {bool([])}")
print(f"bool([1,2]):   {bool([1,2])}")
````

# Part 3: Strings & f-strings (15 min) 📝

Strings are **immutable** sequences of characters.

Key concepts:
- Created with `'single'` or `"double"` quotes
- **Immutable**: you cannot change individual characters
- **f-strings**: the modern way to format output — `f"text {variable}"`
- Useful methods: `.upper()`, `.lower()`, `.title()`, `.startswith()`, `.replace()`, `.split()`

````python
# EXAMPLE: String operations

name = "alice smith"

print(f"Original:     '{name}'")
print(f".upper():     '{name.upper()}'")
print(f".title():     '{name.title()}'")
print(f".replace():   '{name.replace('alice', 'bob')}'")
print(f".split():     {name.split()}")
print(f"len():        {len(name)}")
print(f".startswith('a'): {name.startswith('a')}")

# Concatenation
greeting = "Hello, " + name.title() + "!"
print(f"\nConcatenation: {greeting}")

# Repetition
print(f"Repetition:    {'=-' * 15}")
````

````python
# EXAMPLE: Strings are IMMUTABLE

word = "Hello"

# This would cause an error:
# word[0] = "J"  # TypeError: 'str' object does not support item assignment

# Instead, create a NEW string:
new_word = "J" + word[1:]
print(f"Original: {word}")
print(f"New:      {new_word}")
print(f"Original unchanged? {word == 'Hello'}")
````

# Part 4: Lists (15 min) 📚

Lists are **mutable**, **ordered** collections that can hold any type.

| Operation | Syntax | What It Does |
|-----------|--------|--------------|
| Create | `[1, 2, 3]` | Makes a new list |
| Append | `lst.append(x)` | Add to end |
| Insert | `lst.insert(i, x)` | Add at position i |
| Remove | `lst.remove(x)` | Remove first occurrence |
| Pop | `lst.pop()` | Remove & return last |
| Sort | `lst.sort()` | Sort in place |
| Length | `len(lst)` | Number of elements |

> ⚠️ **Key concept:** Lists are **mutable** — changes happen *in place*. Assigning one list to another variable creates a **reference**, not a copy!

````python
# EXAMPLE: List operations

scores = [85, 92, 78, 95, 88]
print(f"Original:  {scores}")

scores.append(91)
print(f"append(91): {scores}")

scores.sort()
print(f"sort():     {scores}")

scores.reverse()
print(f"reverse():  {scores}")

removed = scores.pop()
print(f"pop():      {scores}  (removed {removed})")

print(f"\nLength: {len(scores)}")
print(f"Min: {min(scores)}, Max: {max(scores)}, Sum: {sum(scores)}")
print(f"Average: {sum(scores) / len(scores):.1f}")
````

````python
# EXAMPLE: Mutable gotcha!

a = [1, 2, 3]
b = a          # b points to the SAME list!
b.append(4)

print(f"a = {a}")  # a changed too!
print(f"b = {b}")
print(f"Same object? {a is b}")

print("\n--- Making a proper copy ---")
c = [1, 2, 3]
d = c.copy()   # d is an INDEPENDENT copy
d.append(4)

print(f"c = {c}")  # c is unchanged
print(f"d = {d}")
print(f"Same object? {c is d}")
````

# Part 5: Tuples & Dictionaries (15 min) 🗂️

### Tuples: Immutable Lists
- Created with `()` or just commas
- **Cannot be changed** after creation
- Use when data should stay fixed (coordinates, RGB colors, multiple return values)

### Dictionaries: Key-Value Pairs
- Created with `{key: value}`
- **Mutable** — you can add, change, and remove entries
- Keys must be immutable (strings, numbers, tuples)
- Perfect for structured data, JSON, and lookup tables

````python
# EXAMPLE: Tuples

# Creating tuples
point = (3, 4)
rgb_red = (255, 0, 0)
single = (42,)    # Single element needs trailing comma!
not_a_tuple = (42) # This is just an integer

print(f"point: {point}, type: {type(point)}")
print(f"single: {single}, type: {type(single)}")
print(f"not_a_tuple: {not_a_tuple}, type: {type(not_a_tuple)}")

# Unpacking
x, y = point
print(f"\nUnpacked: x={x}, y={y}")

# Can't modify!
# point[0] = 10  # This would cause a TypeError
````

````python
# EXAMPLE: Dictionaries

student = {
    "name": "Alice",
    "age": 20,
    "major": "Data Analytics",
    "gpa": 3.85
}

# Access values
print(f"Name: {student['name']}")
print(f"GPA:  {student['gpa']}")

# Safe access with .get() (returns default if key missing)
print(f"Email: {student.get('email', 'Not provided')}")

# Add / modify
student["email"] = "alice@bryant.edu"
student["age"] = 21

# Iterate
print("\n--- Full Profile ---")
for key, value in student.items():
    print(f"  {key}: {value}")
````

# Part 6: Indexing & Slicing (10 min) ✂️

Access individual elements with **indexing** and extract portions with **slicing**.

```
sequence[start : stop : step]
```

- **start** = inclusive (included)
- **stop** = exclusive (NOT included)
- **step** = skip count

```
word = "PYTHON"
         P    Y    T    H    O    N
index:   0    1    2    3    4    5
neg:    -6   -5   -4   -3   -2   -1
```

````python
# EXAMPLE: Indexing and Slicing

word = "PYTHON"
nums = [10, 20, 30, 40, 50, 60, 70, 80]

print("=== Indexing ===")
print(f"word[0]:   '{word[0]}'")
print(f"word[-1]:  '{word[-1]}'")
print(f"nums[2]:   {nums[2]}")

print("\n=== Basic Slicing ===")
print(f"word[0:3]:   '{word[0:3]}'")
print(f"word[2:]:    '{word[2:]}'")
print(f"word[:4]:    '{word[:4]}'")
print(f"nums[-3:]:   {nums[-3:]}")
print(f"nums[:-2]:   {nums[:-2]}")

print("\n=== Step Slicing ===")
print(f"nums[::2]:   {nums[::2]}")
print(f"nums[::-1]:  {nums[::-1]}")
print(f"word[::-1]:  '{word[::-1]}'")
````

### 📝 ACTIVITY 6.1: Slicing Challenge

Use indexing and slicing to extract the requested values. **Do NOT hardcode the values** — use slicing syntax!

````python
# ACTIVITY 6.1: Slicing Challenge

data = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
message = "Data Analytics is awesome!"

# 1. Get the first element of data
first = data[0]      # Expected: "Mon"
print(f"1. First day: {first}")

# 2. Get the last element of data (use negative indexing)
last = data[-1]       # Expected: "Sun"
print(f"2. Last day: {last}")

# 3. Get the weekdays only (Mon-Fri)
weekdays = data[:5]   # Expected: ["Mon", "Tue", "Wed", "Thu", "Fri"]
print(f"3. Weekdays: {weekdays}")

# 4. Get the weekend only (Sat-Sun)
weekend = data[5:]    # Expected: ["Sat", "Sun"]
print(f"4. Weekend: {weekend}")

# 5. Get every other day starting from Monday
every_other = data[::2] # Expected: ["Mon", "Wed", "Fri", "Sun"]
print(f"5. Every other: {every_other}")

# 6. Reverse the list
reversed_days = data[::-1] # Expected: ["Sun", "Sat", "Fri", ...]
print(f"6. Reversed: {reversed_days}")

# 7. Extract just "Analytics" from the message string
analytics = message[5:14]    # Hint: count the characters, use message[start:stop]
print(f"7. Extracted: '{analytics}'")
````
